```python
import yfinance as yf
import pandas as pd

# Download Apple 1-hour data
df = yf.download(
    tickers="AAPL",
    interval="1h",
    period="60d",      # last 60 days (safe for Yahoo hourly)
    auto_adjust=False,
    progress=False
)

df.head()
```

```python
df.to_csv("../data/raw/AAPL_1h.csv")
```
