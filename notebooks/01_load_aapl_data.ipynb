```python
import yfinance as yf
import pandas as pd

# Download Apple 1-hour data
df = yf.download(
    tickers="AAPL",
    interval="1h",
    period="60d",
    auto_adjust=False,
    progress=False
)

# Flatten the MultiIndex columns immediately
df.columns = df.columns.get_level_values(0)

# Reset index so Datetime becomes a normal column
df = df.reset_index()

```

```python
df.to_csv("../data/raw/AAPL_1h.csv", index=False)
```

```python
import pandas as pd

# Read CSV with the correct header row (Datetime row)
df = pd.read_csv("../data/raw/AAPL_1h.csv", header=[0,1], index_col=0)

# Convert index to datetime
df.index = pd.to_datetime(df.index)

df.head()
```
